```python
class Student:
    def __init__(self, name, roll, course):
        self.name = name
        self.roll = roll
        self.course = course

    def display_student(self):
        print("\n----- Student Details -----")
        print("Name :", self.name)
        print("Roll No :", self.roll)
        print("Course :", self.course)


class Teacher:
    def __init__(self, name, subject, salary):
        self.name = name
        self.subject = subject
        self.salary = salary

    def display_teacher(self):
        print("\n----- Teacher Details -----")
        print("Name :", self.name)
        print("Subject :", self.subject)
        print("Salary :", self.salary)


# Main Program
print("===== Student Management =====")
s_name = input("Enter Student Name: ")
s_roll = input("Enter Roll No: ")
s_course = input("Enter Course: ")

student = Student(s_name, s_roll, s_course)
student.display_student()

print("\n===== Teacher Management =====")
t_name = input("Enter Teacher Name: ")
t_subject = input("Enter Subject: ")
t_salary = input("Enter Salary: ")

teacher = Teacher(t_name, t_subject, t_salary)
teacher.display_teacher()
```

```python
# Parent Class
class Car:
    def __init__(self, brand):
        self.brand = brand

    def show(self):
        print("Car Brand:", self.brand)


# Child Class
class ElectricCar(Car):
    def __init__(self, brand, battery):
        super().__init__(brand)
        self.battery = battery

    def display(self):
        self.show()
        print("Battery:", self.battery)


# Main Program
brand = input("Enter Car Brand: ")
battery = input("Enter Battery: ")

obj = ElectricCar(brand, battery)
obj.display()
```
