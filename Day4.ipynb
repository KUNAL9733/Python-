```python
#loops
#if statement--- if our condition is true.
age = 20
if age >= 18:
    print("eligible to vote")
    
```

```python
#if else statement--- if our condition is false.
age = int(input("Enter your age: "))
if age >= 18:
    print("eligible to vote")
else:
    print("not eligible to vote")
```

```python
#if elif else statement--- if our condition is false and we have multiple conditions to check.
marks = int(input("enter your marks:"))
if marks >= 90:
    print("A grade")
elif marks >= 80:
    print("B grade")
else:
    print("C grade")
```

```python
#nested if statement--- if we have multiple conditions to check and one condition is dependent on another condition.
age = int(input("enter your age:"))
citizen = input("are you a citizen? (yes/no): ")
if age >=18 :
    if citizen == "yes":
        print("can vote")
    else:
        print("cannot vote")
else:
    print("cannot vote")    
```
