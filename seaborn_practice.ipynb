```python
import matplotlib
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

print('Seaborn version:', sns.__version__)
```

```python
penguins = sns.load_dataset('penguins')
print('\nPenguins dataset:')
print(penguins.head())
```

```python
diamonds = sns.load_dataset('diamonds').sample(300, random_state=1)
print('\nDiamonds sample:')
print(diamonds.head())
```

```python
planets = sns.load_dataset('planets')
print('\nPlanets dataset:')
print(planets.head())
```

```python
styles = ['white', 'dark', 'whitegrid', 'darkgrid', 'ticks']
fig, axes = plt.subplots(len(styles), 1, figsize=(10, 15))
for i, style in enumerate(styles):
    with sns.axes_style(style):
        vals = np.random.randn(100)
        axes[i].hist(vals, bins=20, color='indianred')
        axes[i].set_title(f'Style: {style}')
plt.tight_layout()
plt.show()
```

```python
contexts = ['paper', 'notebook', 'talk', 'poster']
fig, axes = plt.subplots(len(contexts), 1, figsize=(10, 12))
for i, context in enumerate(contexts):
    with sns.plotting_context(context):
        vals = np.random.randn(100)
        axes[i].hist(vals, bins=20, color='mediumpurple')
        axes[i].set_title(f'Context: {context}')
plt.tight_layout()
plt.show()
```

```python
print('Built-in palettes:')
for palette in ['bright', 'muted', 'pastel', 'dark', 'colorblind']:
    sns.set_palette(palette)
    plt.figure(figsize=(6, 1))
    sns.palplot(sns.color_palette())
    plt.title(palette)
    plt.show()
```

```python
print('\nCustom palettes:')
plt.figure(figsize=(8, 2))
sns.palplot(sns.color_palette('mako', 10))
plt.title('mako')
plt.show()

plt.figure(figsize=(8, 2))
sns.palplot(sns.diverging_palette(220, 20, n=9))
plt.title('diverging')
plt.show()

plt.figure(figsize=(8, 2))
sns.palplot(sns.light_palette('seagreen', 8))
plt.title('light palette')
plt.show()
```

```python
sns.set_theme()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(penguins['body_mass_g'], bins=30, kde=True, color='teal')
plt.title('Histogram of Penguin Body Mass')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.kdeplot(data=penguins, x='flipper_length_mm', hue='species', fill=True, alpha=0.5)
plt.title('KDE of Flipper Length by Species')
plt.show()
```

```python
# ECDF plot
plt.figure(figsize=(10, 6))
sns.ecdfplot(data=penguins, x='body_mass_g', hue='sex')
plt.title('ECDF of Body Mass by Sex')
plt.show()
```

```python
# Box plot
plt.figure(figsize=(10, 6))
sns.boxplot(data=penguins, x='island', y='body_mass_g', hue='sex', palette='Set3')
plt.title('Box Plot of Body Mass by Island and Sex')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.violinplot(data=penguins, x='species', y='bill_length_mm', hue='sex', split=True, palette='Set3')
plt.title('Violin Plot of Bill Length by Species and Sex')
plt.show()
```

```python
# Strip + point plot
plt.figure(figsize=(10, 6))
sns.stripplot(data=penguins, x='species', y='bill_depth_mm', jitter=True, alpha=0.6, color='gray')
sns.pointplot(data=penguins, x='species', y='bill_depth_mm', color='crimson', errorbar=None)
plt.title('Strip Plot with Mean Line')
plt.show()
```

```python
# Rug plot
plt.figure(figsize=(10, 4))
sns.kdeplot(data=penguins, x='flipper_length_mm', color='steelblue')
sns.rugplot(data=penguins, x='flipper_length_mm', color='black')
plt.title('KDE with Rug Plot')
plt.show()
```

```python
# Joint plot
sns.jointplot(data=penguins, x='bill_length_mm', y='bill_depth_mm', kind='scatter', hue='species', height=8)
plt.show()
```

```python
# Pair plot (Diamonds sample)
sns.pairplot(diamonds, vars=['carat', 'depth', 'table', 'price'], hue='cut', palette='husl', height=2.5)
plt.show()
```

```python
# Bar plot with confidence intervals
plt.figure(figsize=(10, 6))
sns.barplot(data=penguins, x='island', y='body_mass_g', hue='sex', palette='Set2', errorbar='sd')
plt.title('Average Body Mass by Island and Sex')
plt.show()
```

```python
# Count plot
plt.figure(figsize=(10, 6))
sns.countplot(data=penguins, x='species', hue='island', palette='pastel')
plt.title('Count of Penguins by Species and Island')
plt.show()
```

```python
# Boxen plot (letter-value)
plt.figure(figsize=(10, 6))
sns.boxenplot(data=diamonds, x='cut', y='price', palette='crest')
plt.title('Boxen Plot of Diamond Price by Cut')
plt.show()
```

```python
# Catplot with different kinds
g = sns.catplot(data=penguins, x='species', y='bill_length_mm', hue='sex', kind='bar', height=5, aspect=1.3, palette='Set2')
g.fig.suptitle('Catplot (bar kind)', y=1.03)
plt.show()
```

```python
# Scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(data=penguins, x='bill_length_mm', y='body_mass_g', hue='species', style='sex', size='flipper_length_mm', sizes=(20, 200), palette='Set2', alpha=0.7)
plt.title('Scatter Plot: Bill Length vs Body Mass')
plt.show()
```

```python
dowjones = sns.load_dataset('dowjones')
plt.figure(figsize=(10, 6))
sns.lineplot(data=dowjones, x='Date', y='Price', color='darkorange')
plt.title('Line Plot: Dow Jones Price Over Time')
plt.show()
```

```python
# Relplot with facets
g = sns.relplot(data=penguins, x='bill_length_mm', y='bill_depth_mm', hue='species', col='sex', kind='scatter', height=4)
g.fig.suptitle('Relplot with Facets by Sex', y=1.03)
plt.show()
```

```python
mpg = sns.load_dataset('mpg').dropna()

# Simple linear regression
plt.figure(figsize=(10, 6))
sns.regplot(data=mpg, x='horsepower', y='mpg', color='steelblue')
plt.title('Linear Regression: Horsepower vs MPG')
plt.show()
```

```python
# Polynomial regression (order=2)
plt.figure(figsize=(10, 6))
sns.regplot(data=mpg, x='horsepower', y='mpg', order=2, color='coral')
plt.title('Polynomial Regression (Order 2)')
plt.show()
```

```python
# Logarithmic regression
plt.figure(figsize=(10, 6))
sns.regplot(data=mpg, x='horsepower', y='mpg', logx=True, color='green')
plt.title('Logarithmic Regression')
plt.show()
```

```python
# Residual plot
plt.figure(figsize=(10, 6))
sns.residplot(data=mpg, x='horsepower', y='mpg', color='purple')
plt.title('Residual Plot')
plt.show()
```

```python
# lmplot with multiple categories
g = sns.lmplot(data=mpg, x='horsepower', y='mpg', hue='origin', height=6, aspect=1.3)
g.fig.suptitle('lmplot by Origin', y=1.03)
plt.show()
```

```python
car_crashes = sns.load_dataset('car_crashes')
plt.figure(figsize=(8, 6))
corr = car_crashes.select_dtypes(include=[np.number]).corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0, square=True, linewidths=0.5)
plt.title('Correlation Heatmap (Car Crashes)')
plt.show()
```

```python
# Flights pivot heatmap
flights = sns.load_dataset('flights')
flights_pivot = flights.pivot(index='month', columns='year', values='passengers')
plt.figure(figsize=(10, 7))
sns.heatmap(flights_pivot, annot=True, fmt='d', cmap='mako', linewidths=0.5)
plt.title('Flight Passengers Heatmap')
plt.show()
```

```python
# Clustered heatmap
sns.clustermap(flights_pivot, cmap='YlGnBu', figsize=(8, 8), annot=False)
plt.show()
```

```python
# FacetGrid
g = sns.FacetGrid(penguins, col='sex', row='island', height=3, aspect=1.2)
g.map(sns.histplot, 'body_mass_g', kde=True)
g.set_axis_labels('Body Mass (g)', 'Count')
g.set_titles(col_template='{col_name}', row_template='Island: {row_name}')
plt.show()
```

```python
# Custom PairGrid
g = sns.PairGrid(penguins.dropna(), vars=['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm'], hue='species')
g.map_upper(sns.scatterplot, alpha=0.6)
g.map_lower(sns.kdeplot)
g.map_diag(sns.histplot, kde=True)
g.add_legend()
plt.show()
```

```python
# Custom JointGrid
g = sns.JointGrid(data=penguins, x='flipper_length_mm', y='body_mass_g', height=7)
g.plot_joint(sns.scatterplot, hue=penguins['species'], alpha=0.7)
g.plot_marginals(sns.histplot, kde=True, color='slateblue')
plt.show()
```

```python
# Comprehensive distribution analysis: histogram, KDE, box plot and violin plot for body_mass_g
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

sns.histplot(penguins['body_mass_g'], kde=False, color='steelblue', ax=axs[0, 0])
axs[0, 0].set_title('Histogram')

sns.kdeplot(data=penguins, x='body_mass_g', hue='species', fill=True, ax=axs[0, 1])
axs[0, 1].set_title('KDE by Species')

sns.boxplot(data=penguins, x='species', y='body_mass_g', palette='Set2', ax=axs[1, 0])
axs[1, 0].set_title('Box Plot')

sns.violinplot(data=penguins, x='species', y='body_mass_g', palette='Set2', ax=axs[1, 1])
axs[1, 1].set_title('Violin Plot')

fig.suptitle('Comprehensive Distribution Analysis - Body Mass', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Correlation and regression dashboard for the penguins dataset
fig, axs = plt.subplots(1, 2, figsize=(14, 6))

num_cols = penguins.select_dtypes(include=[np.number]).corr()
sns.heatmap(num_cols, annot=True, fmt='.2f', cmap='vlag', center=0, ax=axs[0])
axs[0].set_title('Correlation Heatmap')

sns.regplot(data=penguins, x='flipper_length_mm', y='body_mass_g', color='darkgreen', ax=axs[1])
axs[1].set_title('Regression: Flipper Length vs Body Mass')

fig.suptitle('Correlation and Regression Dashboard', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```
