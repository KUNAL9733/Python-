```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np

print('Matplotlib version:', matplotlib.__version__)
print('NumPy version:', np.__version__)
```

```python
# quick line plot to warm up
days = [1, 2, 3, 4, 5, 6, 7]
temp = [22, 24, 23, 26, 28, 27, 25]
plt.plot(days, temp)
plt.show()
```

```python
t = np.linspace(0, 12, 60)

plt.figure(figsize=(10, 6))
plt.plot(t, np.sin(t), label='sin(t)')
plt.plot(t, np.exp(-t/6), label='decay')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.title('Line Plot with Legend and Grid')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# playing around with format strings
plt.figure(figsize=(12, 4))
plt.plot(t, t, 'c-', label='linear')
plt.plot(t, t**0.5, 'y--', label='sqrt')
plt.plot(t, np.log(t + 1), 'k:', label='log')
plt.plot(t, np.exp(t/12), 'r-.', label='exp growth')
plt.legend()
plt.title('Format String Practice')
plt.show()
```

```python
# more detailed styling with markers
plt.figure(figsize=(10, 6))
plt.plot(t, np.sin(t), color='teal', linestyle='--', linewidth=2,
         marker='^', markersize=6, markerfacecolor='white', label='sin(t)')
plt.plot(t, np.cos(t), color='#8E44AD', linestyle='-', linewidth=3,
         marker='D', markersize=4, alpha=0.6, label='cos(t)')
plt.xlabel('Time', fontsize=12)
plt.ylabel('Value', fontsize=12)
plt.title('Custom Styled Line Plot', fontsize=14, fontweight='bold')
plt.legend(loc='lower left', fontsize=10)
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

```python
# fill between two curves
y1 = np.sin(t)
y2 = np.sin(t) * 0.5

plt.figure(figsize=(12, 6))
plt.plot(t, y1, 'b-', linewidth=2, label='sin(t)')
plt.plot(t, y2, 'g--', linewidth=2, label='0.5*sin(t)')
plt.fill_between(t, y1, y2, alpha=0.25, color='orange', label='gap')
plt.xlabel('Time')
plt.ylabel('Value')
plt.title('Filled Area Between Two Lines')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
np.random.seed(7)

# basic scatter
a = np.random.randn(60)
b = np.random.randn(60)
plt.figure(figsize=(8, 6))
plt.scatter(a, b, color='darkgreen', alpha=0.6, s=45)
plt.xlabel('A')
plt.ylabel('B')
plt.title('Basic Scatter Plot')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
a = np.random.randn(120)
b = np.random.randn(120)
groups = np.random.randint(1, 6, 120)
sizes = np.random.randint(15, 220, 120)

plt.figure(figsize=(8, 6))
sc = plt.scatter(a, b, c=groups, s=sizes, alpha=0.65, cmap='cool', edgecolors='black', linewidth=0.5)
plt.colorbar(sc, label='Cluster')
plt.xlabel('A')
plt.ylabel('B')
plt.title('Scatter Colored and Sized by Groups')
plt.show()
```

```python
# scatter with a trend line, like fitting a mini regression
x = np.linspace(0, 10, 80)
y = -3*x + 40 + np.random.normal(0, 4, 80)

plt.figure(figsize=(8, 6))
plt.scatter(x, y, c=y, cmap='autumn', s=35, alpha=0.7)
plt.plot(x, -3*x + 40, 'k--', label='true line')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter with Trend Line')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# bubble chart for made up product sales
products = ['P1', 'P2', 'P3', 'P4', 'P5']
price = np.array([10, 25, 15, 30, 20])
rating = np.array([3.2, 4.5, 3.9, 4.8, 4.1])
units_sold = np.array([300, 900, 500, 1200, 700])

plt.figure(figsize=(8, 6))
plt.scatter(price, rating, s=units_sold, c=units_sold, cmap='spring', alpha=0.6, edgecolors='black')
for i, name in enumerate(products):
    plt.annotate(name, (price[i], rating[i]), textcoords="offset points", xytext=(0, 10), ha='center')
plt.xlabel('Price')
plt.ylabel('Rating')
plt.title('Bubble Chart - Price vs Rating vs Units Sold')
plt.show()
```

```python
# vertical bar chart of weekly steps
weekdays = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
steps = [8500, 9200, 6700, 10400, 7300]

plt.figure(figsize=(8, 5))
bars = plt.bar(weekdays, steps, color=['#3498DB', '#2ECC71', '#E67E22', '#9B59B6', '#E74C3C'], edgecolor='black')
plt.xlabel('Day')
plt.ylabel('Steps')
plt.title('Vertical Bar Chart - Weekly Steps')
plt.bar_label(bars, padding=3)
plt.show()
```

```python
# horizontal bar chart
plt.figure(figsize=(8, 5))
bars = plt.barh(weekdays, steps, color='mediumseagreen', edgecolor='black')
plt.xlabel('Steps')
plt.ylabel('Day')
plt.title('Horizontal Bar Chart - Weekly Steps')
plt.bar_label(bars, padding=3)
plt.show()
```

```python
# grouped bar chart comparing two months
idx = np.arange(5)
last_month = [60, 72, 55, 80, 66]
this_month = [70, 68, 62, 90, 75]
w = 0.35

plt.figure(figsize=(8, 5))
plt.bar(idx - w/2, last_month, w, label='Last Month', color='lightsteelblue', edgecolor='black')
plt.bar(idx + w/2, this_month, w, label='This Month', color='navajowhite', edgecolor='black')
plt.xlabel('Day')
plt.ylabel('Score')
plt.title('Grouped Bar Chart - Month Comparison')
plt.xticks(idx, weekdays)
plt.legend()
plt.show()
```

```python
# stacked bar chart
reading = [2, 3, 1, 4, 2]
gaming = [3, 2, 4, 1, 3]

plt.figure(figsize=(8, 5))
plt.bar(weekdays, reading, label='Reading (hrs)', color='cornflowerblue', edgecolor='black')
plt.bar(weekdays, gaming, bottom=reading, label='Gaming (hrs)', color='salmon', edgecolor='black')
plt.xlabel('Day')
plt.ylabel('Hours')
plt.title('Stacked Bar Chart - Daily Activities')
plt.legend()
plt.show()
```

```python
# bar chart with error bars
avg_score = [72, 68, 81, 75, 90]
error = [4, 6, 3, 5, 2]

plt.figure(figsize=(8, 5))
plt.bar(weekdays, avg_score, yerr=error, capsize=6, color='plum', edgecolor='black')
plt.xlabel('Day')
plt.ylabel('Average Score')
plt.title('Bar Chart with Error Bars')
plt.show()
```

```python
np.random.seed(21)
sample = np.random.normal(50, 10, 800)

# basic histogram
plt.figure(figsize=(8, 5))
plt.hist(sample, bins=25, color='goldenrod', edgecolor='black', alpha=0.75)
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.title('Basic Histogram')
plt.show()
```

```python
# two overlaid histograms
groupA = np.random.normal(50, 8, 800)
groupB = np.random.normal(65, 10, 800)

plt.figure(figsize=(8, 5))
plt.hist(groupA, bins=25, alpha=0.5, label='Group A', color='dodgerblue')
plt.hist(groupB, bins=25, alpha=0.5, label='Group B', color='crimson')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.title('Overlaid Histograms')
plt.legend()
plt.show()
```

```python
# cumulative histogram
plt.figure(figsize=(8, 5))
plt.hist(sample, bins=25, cumulative=True, color='seagreen', edgecolor='black', alpha=0.75)
plt.xlabel('Score')
plt.ylabel('Cumulative Count')
plt.title('Cumulative Histogram')
plt.show()
```

```python
# 2D histogram / hexbin of two correlated variables
xh = np.random.randn(2000)
yh = xh * 0.6 + np.random.randn(2000) * 0.5

plt.figure(figsize=(8, 6))
plt.hexbin(xh, yh, gridsize=25, cmap='inferno')
plt.colorbar(label='Count')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Hexbin Plot')
plt.show()
```

```python
# basic pie chart of budget split
labels = ['Rent', 'Food', 'Travel', 'Savings', 'Misc']
budget = [35, 20, 15, 20, 10]

plt.figure(figsize=(7, 7))
plt.pie(budget, labels=labels, autopct='%1.1f%%', startangle=90,
        colors=['#F1948A', '#85C1E9', '#F8C471', '#82E0AA', '#D2B4DE'])
plt.title('Monthly Budget Split')
plt.show()
```

```python
# exploded pie chart, highlighting savings
explode = (0, 0, 0, 0.12, 0)
plt.figure(figsize=(7, 7))
plt.pie(budget, explode=explode, labels=labels, autopct='%1.1f%%', startangle=90, shadow=True)
plt.title('Exploded Pie Chart - Savings Highlighted')
plt.show()
```

```python
# donut chart
plt.figure(figsize=(7, 7))
plt.pie(budget, labels=labels, autopct='%1.1f%%', startangle=90,
        wedgeprops=dict(width=0.45, edgecolor='white'))
plt.title('Donut Chart')
plt.show()
```

```python
# 2x2 grid using plt.subplot the older way
plt.figure(figsize=(10, 8))
plt.subplot(2, 2, 1)
plt.plot(t, np.sin(t))
plt.title('Sine')
plt.subplot(2, 2, 2)
plt.plot(t, np.cos(t))
plt.title('Cosine')
plt.subplot(2, 2, 3)
plt.plot(t, t**2)
plt.title('Quadratic')
plt.subplot(2, 2, 4)
plt.plot(t, np.exp(-t/6))
plt.title('Exponential Decay')
plt.tight_layout()
plt.show()
```

```python
# same idea using plt.subplots (cleaner way)
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes[0, 0].plot(t, np.sin(t), color='crimson')
axes[0, 0].set_title('Sine')
axes[0, 1].plot(t, np.tan(t), color='navy')
axes[0, 1].set_title('Tangent')
axes[0, 1].set_ylim(-5, 5)
axes[1, 0].plot(t, np.sqrt(t), color='darkgreen')
axes[1, 0].set_title('Square Root')
axes[1, 1].plot(t, np.log(t + 1), color='chocolate')
axes[1, 1].set_title('Logarithm')
fig.suptitle('Four Functions Together', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
np.random.seed(3)
groups_data = [np.random.normal(loc, 4, 120) for loc in [10, 20, 15, 25]]

# box plot
plt.figure(figsize=(8, 5))
plt.boxplot(groups_data, labels=['G1', 'G2', 'G3', 'G4'], patch_artist=True,
            boxprops=dict(facecolor='lightyellow'))
plt.xlabel('Group')
plt.ylabel('Value')
plt.title('Box Plot')
plt.show()
```

```python
# violin plot for the same groups
plt.figure(figsize=(8, 5))
plt.violinplot(groups_data, showmeans=True, showmedians=True)
plt.xlabel('Group')
plt.ylabel('Value')
plt.title('Violin Plot')
plt.xticks([1, 2, 3, 4], ['G1', 'G2', 'G3', 'G4'])
plt.show()
```

```python
# heatmap of made-up correlation-ish grid
grid = np.random.randint(0, 50, (8, 8))
plt.figure(figsize=(8, 6))
im = plt.imshow(grid, cmap='cool', aspect='auto')
plt.colorbar(im, label='Value')
plt.xlabel('Column')
plt.ylabel('Row')
plt.title('Heatmap')
for i in range(8):
    for j in range(8):
        plt.text(j, i, grid[i, j], ha='center', va='center', color='black', fontsize=8)
plt.show()
```

```python
# contour plot
xc, yc = np.meshgrid(np.linspace(-3, 3, 100), np.linspace(-3, 3, 100))
zc = np.exp(-(xc**2 + yc**2))

plt.figure(figsize=(8, 6))
cp = plt.contourf(xc, yc, zc, levels=20, cmap='viridis')
plt.colorbar(cp, label='Z')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Contour Plot')
plt.show()
```

```python
# 3D surface plot
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
X3, Y3 = np.meshgrid(np.linspace(-4, 4, 60), np.linspace(-4, 4, 60))
Z3 = np.cos(np.sqrt(X3**2 + Y3**2))
ax.plot_surface(X3, Y3, Z3, cmap='plasma', edgecolor='none')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Surface Plot')
plt.show()
```
