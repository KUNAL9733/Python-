```python
#set functions
fruits={"apple","banana","orange"}
fruits.add("grapes")
print(fruits)
fruits.remove("banana")
print(fruits)
print(len(fruits))
```

```python
a={1,2,3}
b={3,4,5}
print(a.intersection(b))
print(a.difference(b))
```

```python
student = {
    "name": "Kunal",
    "age": 20,
    "marks": 85
}
print(student)

#acces elements
print(student["name"])
print(student.get("age"))
student['city'] = "New York"
print(student)
student['marks'] = 90
print(student)
del student["marks"]
print(student)
print(student.keys())
print(student.values())
```
