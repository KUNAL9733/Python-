<h1>Matplotlib in Python</h1>

<h2>Matplotlib is a popular Python library used for data visualization. It helps us create different types of graphs and charts to represent data in a visual form.</h2>

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)
plt.show()
```

<h2>Line Plot</h2>

```python
import matplotlib.pyplot as plt

days = [1,2,3,4,5]
sales = [100,150,120,180,200]

plt.plot(days, sales)

plt.title("Daily Sales")
plt.xlabel("Days")
plt.ylabel("Sales")

plt.show()
```

<h2>Bar Chart</h2>

```python
import matplotlib.pyplot as plt

students = ["Rahul", "Aman", "Priya", "Neha"]
marks = [85, 90, 88, 75]

plt.bar(students, marks)

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

<h2>Horizontal Bar Chart</h2>

```python
import matplotlib.pyplot as plt

students = ["Rahul", "Aman", "Priya", "Neha"]
marks = [85, 90, 88, 75]

plt.barh(students, marks)

plt.show()
```

<h2>Pie Chart</h2>

```python
import matplotlib.pyplot as plt

subjects = ["Math", "Science", "English"]
marks = [40, 35, 25]

plt.pie(marks, labels=subjects, autopct="%1.1f%%")

plt.title("Subject Marks")

plt.show()
```

<h2>Scatter Plot</h2>

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,4,5,4,6]

plt.scatter(x, y)

plt.title("Scatter Plot")

plt.show()
```

<h2>Histogram</h2>

```python
import matplotlib.pyplot as plt

marks = [65,70,75,80,85,90,95,80,75,70]

plt.hist(marks)

plt.title("Marks Distribution")

plt.show()
```

<h2>Grid</h2>

```python
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,20,15,25]

plt.plot(x, y)

plt.grid()

plt.show()
```

```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np

print('Matplotlib version:', matplotlib.__version__)
print('NumPy version:', np.__version__)
```

```python
x = [1, 2, 3, 4, 5, 6, 7, 8]
y = [1, 4, 9, 16, 25, 30, 45, 50]
plt.plot(x,y)
plt.show()
```

```python
x = np.linspace(0, 10, 100)
plt.figure(figsize=(8, 6))
plt.plot(x, np.sin(x))
plt.show()
```

```python
x = np.linspace(0, 10, 100)
plt.figure(figsize=(8, 6))
plt.plot(x, np.sin(x), label='sin(x)')
plt.plot(x, np.cos(x), label='cos(x)')
plt.title('waveform')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(15, 7))
plt.plot(x, x, 'r-', label='linear equation')
plt.plot(x, x**2, 'g--', label='quadratic equation')
plt.plot(x, x**3, 'b:', label='cubic root')
plt.plot(x, np.sqrt(x), 'm-.', label='square root')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(x, np.sin(x), color='blue', linestyle='--', linewidth=2, marker='o', markersize=5, markerfacecolor='yellow', label='sin(x)')
plt.plot(x, np.cos(x), color="#24F815", linestyle='-', linewidth=3, marker='s', markersize=4, markerfacecolor='red',alpha=0.7, label='cos(x)')
plt.xlabel('X', fontsize=12)
plt.ylabel('Y', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.7)
plt.show()
```

```python
data = [12, 15, 18, 20, 22, 25, 25, 28, 30, 32, 35, 38, 40]

plt.hist(data, bins=5, density=True )

plt.title("Density Histogram")
plt.xlabel("Values")
plt.ylabel("Density")
plt.show()
```

```python
data = [10, 15, 20, 25, 30, 35, 40, 45, 50]
plt.hist(data, bins=5, cumulative=True)
plt.title("Cumulative Histogram")
plt.xlabel("Values")
plt.ylabel("Cumulative Frequency")
plt.show()
```

```python

data = [10, 15, 20, 25, 30, 35, 40, 45, 50]
plt.hist(data, bins=5, histtype="step", linewidth=2)
plt.title("Step Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.grid(True)

# Display the graph
plt.show()
```

```python


# Sample data
data = [11, 22, 33, 44, 55, 66, 77, 88, 99]
plt.figure(figsize=(6, 4))
plt.hist(data,bins=5,histtype="step",linewidth=2,color="blue",linestyle="--")

plt.title("Step Histogram", fontsize=14)
plt.xlabel("Values", fontsize=12)
plt.ylabel("Frequency", fontsize=12)

plt.grid(True, linestyle=":", alpha=0.7)
plt.show()
```

```python
x = [2,4,6,8,10,12,14,16]
y = [4,8,12,16,20,24,28,32]

plt.hist2d(x, y, bins=4)
plt.title("2D Histogram")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.colorbar(label="Frequency")

plt.grid(True)
plt.show()
```

```python

outer = [40, 30, 20, 10]
inner = [20, 20, 15, 15, 10, 10, 5, 5]

plt.figure(figsize=(6, 6))
plt.pie(
    outer,
    radius=1,
    labels=["A", "B", "C", "D"],
    colors=["red", "blue", "green", "orange"],
    autopct="%1.1f%%"
)
plt.pie(
    inner,
    radius=0.6,
    labels=["A1", "B1", "C1", "D1", "E2"],
    colors=["pink", "lightblue", "lightgreen", "yellow",
             "brown"]
)
plt.title("Nested Pie Chart")
plt.show()
```
