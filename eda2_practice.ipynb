```python
# What is the average price for each room type?
df.groupby('room_type')['price'].mean()
```

```python
# How many hosts have more than 1 listing?
host_counts = df['host_id'].value_counts()
(host_counts > 1).sum()
```

```python
# Which neighbourhood group has the highest average price?
df.groupby('neighbourhood_group')['price'].mean().sort_values(ascending=False)
```

```python
# What is the cheapest listing in the dataset, and what is its name?
cheapest_price = df['price'].min()
df[df['price'] == cheapest_price][['name', 'neighbourhood_group', 'price']]
```

```python
# Which listing has the highest reviews per month?
top_reviewed = df.sort_values('reviews_per_month', ascending=False).iloc[0]
print(top_reviewed)
```

```python
# What is the average minimum nights required, broken down by neighbourhood group?
df.groupby('neighbourhood_group')['minimum_nights'].mean()
```

```python
# How many listings are available for more than 200 days a year?
(df['availability_365'] > 200).sum()
```

```python
# Top 5 neighbourhoods with the most listings
df['neighbourhood'].value_counts().head(5)
```

```python
# Find all listings whose most recent review was in 2019
df['last_review'] = pd.to_datetime(df['last_review'])
reviewed_2019 = df[df['last_review'].dt.year == 2019]
reviewed_2019.shape[0]
```

```python
# Correlation between price and number of reviews
df[['price', 'number_of_reviews']].corr()
```
