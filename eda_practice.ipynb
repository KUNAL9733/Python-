```python
df1 = pd.DataFrame({'EmpID': [101, 102, 103, 104], 'Name': ['Kabir', 'Meera', 'Arjun', 'Divya']})
df2 = pd.DataFrame({'EmpID': [103, 104, 105, 106], 'Salary': [45000, 62000, 51000, 70000]})
```

```python
print(df1)
```

```python
print(df2)
```

```python
print('\n=== Inner Join (default) ===')
print(pd.merge(df1, df2, on='EmpID'))
```

```python
print('\n=== Left Join ===')
print(pd.merge(df1, df2, on='EmpID', how='left'))
```

```python
print('\n=== Right Join ===')
print(pd.merge(df1, df2, on='EmpID', how='right'))
```

```python
print('\n=== Outer Join ===')
print(pd.merge(df1, df2, on='EmpID', how='outer'))
```

```python
df3 = pd.DataFrame({'Emp_Code': [101, 102, 103], 'Grade': ['A', 'B', 'C']})
print(df3)
```

```python
print('\nMerge on different names:')
print(pd.merge(df1, df3, left_on='EmpID', right_on='Emp_Code'))
```

```python
# quick recap: inner, outer, left and right join in one go
for kind in ['inner', 'outer', 'left', 'right']:
    print(f'\n=== {kind.upper()} JOIN ===')
    print(pd.merge(df1, df2, on='EmpID', how=kind))
```

```python
# Load a real dataset for the next part
df = pd.read_csv("neet_data.csv")
```

```python
# Show column information and total rows/columns
print(df.info())
print('\nTotal rows:', df.shape[0])
print('Total columns:', df.shape[1])
```

```python
# Sort the dataset with highest retweet to lowest
df_sorted = df.sort_values('retweet_count', ascending=False)
df_sorted.head()
```

```python
# Show the tweet source, count, user_name, tweet, created_at and user_url for the highest retweet count
top_row = df_sorted.iloc[0]
top_row[['source', 'retweet_count', 'user_name', 'tweet', 'created_at', 'user_url']]
```

```python
# Top 10 retweet count values, printed with a heading
top10_sum = df_sorted.head(10)['retweet_count'].sum()
print(f'TOP 10 Retweet sum is {top10_sum}')
df_sorted.head(10)[['user_name', 'retweet_count']]
```

```python
# Count and sum of retweets done via Twitter for Android
android_tweets = df[df['source'] == 'Twitter for Android']
print('Total tweets from Twitter for Android:', android_tweets.shape[0])
print('Sum of retweet_count from Twitter for Android:', android_tweets['retweet_count'].sum())
```

```python
# Username and location of the person with the highest follower count
top_follower = df.loc[df['user_follower_count'].idxmax()]
print('User:', top_follower['user_name'])
print('Location:', top_follower['user_location'])
```

```python
# New column with follower to friends ratio
df['followers_friend_ratio'] = df['user_follower_count'] / df['user_friends_count']
df[['user_name', 'user_follower_count', 'user_friends_count', 'followers_friend_ratio']].head()
```

```python
# Total null values in each column
df.isnull().sum()
```

```python
# Fill user_description with 'Twitter user' where it is NA
df['user_description'] = df['user_description'].fillna('Twitter user')
df['user_description'].isnull().sum()
```

```python
# Drop all columns that have any null values
df.dropna(axis=1, inplace=True)
df.columns
```

```python
# Create a new column at the 3rd position named is_allow, set to True
df.insert(2, 'is_allow', True)
df.head()
```

```python
# Drop the user_url column (no longer needed for this analysis)
df.drop(columns=['user_url'], inplace=True, errors='ignore')
df.columns
```
