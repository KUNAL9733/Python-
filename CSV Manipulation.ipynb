<h1>CSV Dataset Manipulation in Python (Using Pandas)</h1>

<h2>A CSV (Comma-Separated Values) file is a text file that stores data in rows and columns. Each value is separated by a comma (,). It is one of the most common file formats for storing and exchanging data.</h2>

<h3>Example=></h3>

<h2>Step 1: Import Pandas</h2>

```python
import pandas as pd
```

<h2>Step 2: Read the CSV File</h2>

```python
import pandas as pd

df = pd.read_csv("students.csv")

print(df)
```

<h2>Step 3: Display Basic Information</h2>

```python
print(df.head())      # First 5 rows
print(df.tail())      # Last 5 rows
print(df.shape)       # Number of rows and columns
print(df.columns)     # Column names
print(df.info())      # Dataset information
```

<h2>Step 4: Select Specific Columns</h2>

print(df["Name"])

print(df[["Name", "Marks"]])

<h2>Step 5: Filter Data</h2>

```python
result = df[df["Marks"] > 80]

print(result)
```

<h2>Step 6: Sort Data</h2>

```python
print(df.sort_values(by="Marks"))
```

<h2>Step 7: Add a New Column</h2>

```python
df["Grade"] = ["B", "A", "A", "C"]

print(df)
```

<h2>Step 8: Update Data</h2>

```python
df.loc[0, "Marks"] = 95

print(df)
```

<h2>Step 9: Delete a Column</h2>

```python
df.drop("Grade", axis=1, inplace=True)

print(df)
```

<h2>Step 10: Delete a Row</h2>

```python
df.drop(3, axis=0, inplace=True)

print(df)
```

<h2>Step 11: Find Average Marks</h2>

```python
average = df["Marks"].mean()

print("Average Marks =", average)
```

<h2>Step 12: Find Maximum and Minimum Marks</h2>

print("Highest Marks =", df["Marks"].max())
print("Lowest Marks =", df["Marks"].min())

<h2>Step 13: Check Missing Values</h2>

```python
print(df.isnull())
print(df.isnull().sum())
```

<h2>Step 14: Fill Missing Values</h2>

```python
df.fillna(0, inplace=True)
```

<h2>Step 15: Save the Modified Data</h2>

 df.to_csv("updated_students.csv", index=False)

print("File Saved Successfully")
