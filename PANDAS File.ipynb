<h2>5.Add a New Column</h2>

```python
import pandas as pd

student = {
    "Name": ["Rahul", "Aman", "Priya"],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(student)

df["Grade"] = ["B", "A", "A"]

print(df)
```

<h2>6.Calculate Average Marks</h2>

```python
import pandas as pd

student = {
    "Name": ["Rahul", "Aman", "Priya"],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(student)

average = df["Marks"].mean()

print("Average Marks =", average)
```

<h2>7.Sort Data by Marks</h2>

```python
import pandas as pd

student = {
    "Name": ["Rahul", "Aman", "Priya", "Neha"],
    "Marks": [85, 90, 88, 70]
}

df = pd.DataFrame(student)

print(df.sort_values(by="Marks", ascending=False))
```

<h1>Complete Program for Pandas in One</h1>

```python
import pandas as pd

# Create data
student = {
    "Name": ["Rahul", "Aman", "Priya", "Neha"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 88, 75]
}

# Create DataFrame
df = pd.DataFrame(student)

# Display Data
print("Student Data")
print(df)

# Display first two rows
print("\nFirst Two Rows")
print(df.head(2))

# Display student names
print("\nNames")
print(df["Name"])

# Filter students with marks above 80
print("\nStudents with Marks > 80")
print(df[df["Marks"] > 80])

# Average marks
print("\nAverage Marks")
print(df["Marks"].mean())

# Add Grade column
df["Grade"] = ["B", "A", "A", "C"]

print("\nUpdated Data")
print(df)
```
