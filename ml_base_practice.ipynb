### Regression
Regression is a type of Machine Learning problem where the output is a continuous numerical value.
* Goal: predict a number.
| Problem | Output |
| ------- | ------ |
| Predict rent price | 25,000 |
| Predict tomorrow's temperature | 28.4C |
| Predict delivery time | 35 minutes |

````python
import numpy as np
from sklearn.linear_model import LinearRegression

# Input (years of experience)
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
# Output (monthly salary in thousands)
y = np.array([25, 32, 40, 47, 55])

model = LinearRegression()
model.fit(X, y)

prediction = model.predict([[8]])
print("Predicted Salary (in thousands):", prediction[0])
````

### Classification
Classification is a type of Machine Learning problem where the output is a category or class.
| Problem | Output |
| ------- | ------ |
| Email spam detection | Spam / Not Spam |
| Loan approval | Approved / Rejected |
| Product review | Positive / Negative |

````python
import numpy as np
from sklearn.linear_model import LogisticRegression

# Input (hours of sleep)
X = np.array([3, 4, 5, 6, 7, 8]).reshape(-1, 1)

# Output (0 = Tired, 1 = Fresh)
y = np.array([0, 0, 0, 1, 1, 1])

model = LogisticRegression()
model.fit(X, y)

prediction = model.predict([[5.5]])
print("Prediction:", "Fresh" if prediction[0] == 1 else "Tired")
````

#### Regression vs Classification
| Feature | Regression | Classification |
| ------- | ---------- | --------------- |
| Output | Number | Category |
| Example | Salary = 55K | Fresh / Tired |
| Algorithms | Linear Regression | Logistic Regression |
| Graph | Best-fit line | Decision boundary |

### KNN (K Nearest Neighbor)
    K-Nearest Neighbors (KNN) is a supervised machine learning algorithm used for both classification and regression tasks.
    It predicts a new data point's label based on the labels of its nearest neighbors.

````python
from sklearn.neighbors import KNeighborsClassifier
````

````python
X = [  # weight(g), texture_score
    [150, 6],
    [170, 6.5],
    [200, 7],
    [280, 8],
    [310, 8.5],
    [340, 9]
]
# 0 = Lemon, 1 = Grapefruit
y = [0, 0, 0, 1, 1, 1]
````

````python
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)
````

````python
prediction = model.predict([[250, 7.5]])[0]
print("Prediction:", "Grapefruit" if prediction == 1 else "Lemon")
````

#### Example 1: House price style regression, but for delivery distance vs cost

````python
# Step 1: Load libraries
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
````

````python
# Step 2: Sample data (distance in km and delivery cost)
distance = [[2], [4], [6], [8], [10], [12], [15], [20]]
cost = [30, 45, 55, 70, 85, 95, 115, 140]  # in rupees
````

````python
# Step 3: Train-test split (80% training, 20% testing)
x_train, x_test, y_train, y_test = train_test_split(distance, cost, test_size=0.2, random_state=42)
````

````python
# Step 4: Create and train the model
model = LinearRegression()
model.fit(x_train, y_train)
````

````python
# Step 5: Predict costs on test data
predicted_costs = model.predict(x_test)
````

````python
# Step 6: Output the result
print("Actual Costs:", y_test)
print("Predicted Costs:", predicted_costs)
````

````python
# Step 7: Visualize the regression line
plt.scatter(distance, cost, color='blue', label='Actual data')
plt.plot(distance, model.predict(distance), color='red', label='Regression line')
plt.xlabel("Distance (km)")
plt.ylabel("Delivery Cost (Rs)")
plt.title("Delivery Cost Prediction")
plt.legend()
plt.grid(True)
plt.show()
````

````python
# Step 1: Import required libraries
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import numpy as np

# Step 2: Sample dataset
distance = [[2], [4], [6], [8], [10], [12], [15], [20]]
cost = [30, 45, 55, 70, 85, 95, 115, 140]

# Step 3: Train-test split
x_train, x_test, y_train, y_test = train_test_split(distance, cost, test_size=0.2, random_state=42)

# Step 4: Train model
model = LinearRegression()
model.fit(x_train, y_train)

# Step 5: Predict
y_pred = model.predict(x_test)

# Step 6: Evaluate
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("MSE:", mse)
print("RMSE:", rmse)
print("MAE:", mae)
print("R2 score:", r2)
````

MSE = Mean Squared Error -> average of squared prediction errors
RMSE = Root Mean Squared Error -> same units as the target, easier to interpret
MAE = Mean Absolute Error -> average of absolute prediction errors, less sensitive to outliers than MSE
R2 score = how much of the variance in the target the model explains (closer to 1 is better)
